# Solving with classical heuristics

Several classical heuristics are made available for solving QUBOs such as tabu search or simulated annealing.

## Quickstart: Solving a QUBO with Tabu Search

Below, we show how to solve a QUBO problem using the Tabu Search solver.


```python
from qubosolver import Instance, Solver, SolverConfig, ClassicalConfig, matrix

qubo = Instance(matrix=matrix.tensor([[-2.0, 1.0], [1.0, -2.0]]))

# we can return at most max_bitstrings solutions.
config = SolverConfig(use_quantum = False, classical=ClassicalConfig(classical_solver_type="tabu_search", max_bitstrings=2))

solver = Solver(qubo, config)

solution = solver.solve()
print(solution)
```

## Quickstart: Solving a QUBO with Simulated Annealing

Below, we show how to solve a QUBO problem using the Simulated Annealing solver.

```python
from qubosolver import Instance, Solver, SolverConfig, ClassicalConfig, matrix

qubo = Instance(matrix=matrix.tensor([[-2.0, 1.0], [1.0, -2.0]]))
config = SolverConfig(
    use_quantum=False,
    classical=ClassicalConfig(
        classical_solver_type="simulated_annealing",
        max_bitstrings=2,
        max_iter=1000,
        sa_time_limit=300,
    ),
)

solver = Solver(qubo, config)

solution = solver.solve()
print(solution)
```

## Quickstart: Solving a QUBO with Simulated Annealing followed by tabu search

There is no single `classical_solver_type` for chaining Simulated Annealing into Tabu Search;
instead, compose the two functional solvers directly: run `solvers.simulated_annealing` to get
a starting bitstring, then feed it into `solvers.tabu_search`.

```python
from qubosolver import Instance, matrix, solvers, torch_rng

qubo = Instance(matrix=matrix.tensor([[-2.0, 1.0], [1.0, -2.0]]))
rng = torch_rng(seed=0)

sa_solution = solvers.simulated_annealing(qubo, solvers.random_solutions(qubo, rng=rng).bitstrings[0], top_k=2)
solution = solvers.tabu_search(qubo, sa_solution.bitstrings, time_limit=300.0)
print(solution)
```
