# Getting started

`qubo-solver` solves Quadratic Unconstrained Binary Optimization (QUBO) instances using either classical or quantum approaches. Here is the entire quantum workflow: define an `Instance` from a QUBO matrix, map its variables onto atoms (embedding), build the laser drive (drive shaping), then compile and run the program on a backend.

The example below runs on a local emulator. See the next tutorial for how to use other quantum backends and classical solvers.

```python
from __future__ import annotations

import numpy as np
import qoolqit

from qubosolver import (
    Instance,
    LocalEmulator,
    Solution,
    analysis,
    drive_shaping,
    embedding,
    solving,
)

# Define QUBO
Q = np.asarray(
    [
        [-0.2, 0.0, 1.0],
        [0.0, -1.0, 1.5],
        [1.0, 1.5, -0.1],
    ],
    dtype=np.float64,
)
instance = Instance(Q)

# Solve it on a local emulator
device = qoolqit.AnalogDeviceWithDMM()
backend = LocalEmulator()

register = embedding.blade.embed_for_device(instance, device)
drive = drive_shaping.local_energy_scale.build_drive(instance, register, device=device)
program = solving.analog_quantum_sampling.compile(register, drive, device)
job = backend.run(program)
solution = Solution.from_results(job.results(), instance)

print(analysis.to_dataframe([solution]))
```

The output is a `Solution`, converted here to a dataframe listing each sampled bitstring, its cost, and its probability. Continue to the next tutorial, [Solving a QUBO problem easily](01-qubosolver-basics.ipynb), for quantum backends and classical solvers.
